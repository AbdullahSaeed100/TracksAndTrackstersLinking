```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import torch
from IPython.display import display


# ============================================================
# 1. Load the real checkpoint
# ============================================================

CHECKPOINT_PATH = Path(
    "/eos/home-i01/a/aasiri/SWAN_projects/Trk/full/checkpoints/"
    "weighted_bce_stratified_bs4_bfc027bbcab5_best.pt"
)

if not CHECKPOINT_PATH.exists():
    raise FileNotFoundError(f"Checkpoint not found:\n{CHECKPOINT_PATH}")

# weights_only=False is required because this trusted checkpoint contains
# optimizer state, metadata, and random-number-generator state.
try:
    checkpoint = torch.load(
        CHECKPOINT_PATH,
        map_location="cpu",
        weights_only=False,
    )
except TypeError:
    # Compatibility with older PyTorch versions.
    checkpoint = torch.load(
        CHECKPOINT_PATH,
        map_location="cpu",
    )

metadata = checkpoint["metadata"]
training_config = metadata["training_config"]
validation = checkpoint["validation_results"]
training_at_best_epoch = checkpoint["training_results"]
early_stopping = checkpoint["early_stopping"]

best_epoch = int(checkpoint["epoch"]) + 1


# ============================================================
# 2. General checkpoint information
# ============================================================

print("=" * 72)
print("                 FRESH STRATIFIED BASELINE")
print("=" * 72)
print(f"Checkpoint loaded successfully: {CHECKPOINT_PATH.name}")
print(f"Checkpoint format version:      {checkpoint['format_version']}")
print(f"Best epoch:                     {best_epoch}")
print(f"Split fingerprint:              {metadata['split_fingerprint']}")
print(
    "Preprocessing fingerprint:     "
    f"{metadata['preprocessing_fingerprint']}"
)


# ============================================================
# 3. Training configuration from checkpoint
# ============================================================

optimizer_config = training_config["optimizer"]
stopping_config = training_config["early_stopping"]
model_config = metadata["model_config"]

configuration = pd.DataFrame(
    {
        "Parameter": [
            "Random seed",
            "Batch size",
            "Optimizer",
            "Learning rate",
            "Weight decay",
            "Epoch budget",
            "Early-stopping patience",
            "Early-stopping minimum delta",
            "Loss type",
            "trk-ts task weight",
            "ts-ts task weight",
            "Number of workers",
        ],
        "Value": [
            metadata["seed"],
            training_config["batch_size"],
            optimizer_config["name"],
            optimizer_config["learning_rate"],
            optimizer_config["weight_decay"],
            training_config["epoch_budget"],
            stopping_config["patience"],
            stopping_config["min_delta"],
            training_config["loss_type"],
            training_config["task_weights"]["trk_ts"],
            training_config["task_weights"]["ts_ts"],
            training_config["num_workers"],
        ],
    }
)

print("\nTRAINING CONFIGURATION")
display(configuration)


# ============================================================
# 4. Model configuration from checkpoint
# ============================================================

model_configuration = pd.DataFrame(
    {
        "Model parameter": list(model_config.keys()),
        "Value": list(model_config.values()),
    }
)

print("\nMODEL CONFIGURATION")
display(model_configuration)


# ============================================================
# 5. Losses saved at the best epoch
# ============================================================

loss_summary = pd.DataFrame(
    {
        "Dataset": ["Training", "Validation"],
        "trk-ts loss": [
            training_at_best_epoch["loss"]["trk_ts"],
            validation["loss"]["trk_ts"],
        ],
        "ts-ts loss": [
            training_at_best_epoch["loss"]["ts_ts"],
            validation["loss"]["ts_ts"],
        ],
        "Combined loss": [
            training_at_best_epoch["loss"]["combined"],
            validation["loss"]["combined"],
        ],
    }
)

print(f"\nLOSSES AT BEST EPOCH {best_epoch}")
display(loss_summary.round(6))


# ============================================================
# 6. Validation dataset composition from checkpoint
# ============================================================

validation_composition = pd.DataFrame(
    {
        "Edge type": ["trk-ts", "ts-ts"],
        "Validation edges": [
            validation["trk_ts"]["edge_count"],
            validation["ts_ts"]["edge_count"],
        ],
        "Positive edges": [
            validation["trk_ts"]["positive_count"],
            validation["ts_ts"]["positive_count"],
        ],
        "Positive fraction": [
            validation["trk_ts"]["positive_fraction"],
            validation["ts_ts"]["positive_fraction"],
        ],
        "Positive percentage": [
            100 * validation["trk_ts"]["positive_fraction"],
            100 * validation["ts_ts"]["positive_fraction"],
        ],
    }
)

print("\nVALIDATION DATA COMPOSITION")
display(validation_composition.round(6))


# ============================================================
# 7. Ranking metrics from checkpoint
# ============================================================

ranking_rows = []

for edge_key, edge_label in [
    ("trk_ts", "trk-ts"),
    ("ts_ts", "ts-ts"),
]:
    edge_results = validation[edge_key]
    ranking = edge_results["ranking"]
    positive_fraction = edge_results["positive_fraction"]

    ranking_rows.append(
        {
            "Edge type": edge_label,
            "Average precision": ranking["average_precision"],
            "ROC-AUC": ranking["roc_auc"],
            "Random AP baseline": positive_fraction,
            "AP lift over random": (
                ranking["average_precision"] / positive_fraction
            ),
        }
    )

ranking_metrics = pd.DataFrame(ranking_rows)

print("\nBEST VALIDATION RANKING METRICS")
display(ranking_metrics.round(6))


# ============================================================
# 8. Best-F1 diagnostics from checkpoint
# ============================================================

diagnostic_rows = []

for edge_key, edge_label in [
    ("trk_ts", "trk-ts"),
    ("ts_ts", "ts-ts"),
]:
    diagnostic = validation[edge_key]["best_f1"]

    diagnostic_rows.append(
        {
            "Edge type": edge_label,
            "Threshold": diagnostic["threshold"],
            "Precision": diagnostic["precision"],
            "Recall": diagnostic["recall"],
            "F1": diagnostic["f1"],
            "Specificity": diagnostic["specificity"],
            "False-positive rate": diagnostic["false_positive_rate"],
            "Balanced accuracy": diagnostic["balanced_accuracy"],
            "Accuracy": diagnostic["accuracy"],
        }
    )

diagnostic_table = pd.DataFrame(diagnostic_rows)

print("\nBEST-F1 VALIDATION DIAGNOSTICS")
display(diagnostic_table.round(6))


# ============================================================
# 9. Confusion-matrix counts from checkpoint
# ============================================================

confusion_rows = []

for edge_key, edge_label in [
    ("trk_ts", "trk-ts"),
    ("ts_ts", "ts-ts"),
]:
    diagnostic = validation[edge_key]["best_f1"]

    confusion_rows.append(
        {
            "Edge type": edge_label,
            "True negative": diagnostic["tn"],
            "False positive": diagnostic["fp"],
            "False negative": diagnostic["fn"],
            "True positive": diagnostic["tp"],
        }
    )

confusion_table = pd.DataFrame(confusion_rows)

print("\nCONFUSION-MATRIX COUNTS")
display(confusion_table)


# ============================================================
# 10. Draw the two conventional confusion matrices
#
# Layout:
#
#                   Predicted
#                 Negative  Positive
# Actual Negative     TN        FP
# Actual Positive     FN        TP
# ============================================================

sns.set_theme(style="white")

fig, axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(12, 5),
)

for axis, edge_key, edge_label in zip(
    axes,
    ["trk_ts", "ts_ts"],
    ["trk-ts", "ts-ts"],
):
    diagnostic = validation[edge_key]["best_f1"]

    confusion_matrix = [
        [diagnostic["tn"], diagnostic["fp"]],
        [diagnostic["fn"], diagnostic["tp"]],
    ]

    sns.heatmap(
        confusion_matrix,
        annot=True,
        fmt=",",
        cmap="Blues",
        cbar=False,
        linewidths=1,
        linecolor="white",
        square=True,
        xticklabels=["Negative", "Positive"],
        yticklabels=["Negative", "Positive"],
        annot_kws={"fontsize": 13},
        ax=axis,
    )

    axis.set_title(
        f"{edge_label} confusion matrix\n"
        f"Best-F1 threshold = {diagnostic['threshold']:.4f}",
        fontsize=13,
        pad=12,
    )
    axis.set_xlabel("Predicted class", fontsize=11)
    axis.set_ylabel("Actual class", fontsize=11)
    axis.tick_params(axis="x", rotation=0)
    axis.tick_params(axis="y", rotation=0)

plt.suptitle(
    f"Validation Confusion Matrices — Best Epoch {best_epoch}",
    fontsize=15,
    y=1.03,
)

plt.tight_layout()
plt.show()


# ============================================================
# 11. Reproducibility information
# ============================================================

reproducibility = pd.DataFrame(
    {
        "Item": [
            "Checkpoint path",
            "Split fingerprint",
            "Preprocessing fingerprint",
        ],
        "Value": [
            str(CHECKPOINT_PATH),
            metadata["split_fingerprint"],
            metadata["preprocessing_fingerprint"],
        ],
    }
)

print("\nREPRODUCIBILITY INFORMATION")
display(reproducibility)

print("\n" + "=" * 72)
print("CHECKPOINT REPORT COMPLETED")
print("=" * 72)
```
